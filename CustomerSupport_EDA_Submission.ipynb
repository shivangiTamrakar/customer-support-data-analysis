## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
#1. check dataset shape
print("Dataset Shape:", df.shape)

#2.check data types
print("Data types:")
print(df.dtypes)

#3.check missing values
print("Missing Values:",df.isnull().sum())

#4.check duplicate records
print("Duplicate Rows:",df.duplicated().sum())

#5. Remove duplicate rows(if any)
df = df.drop_duplicates()

#6. Verify duplicates after removal
print("Duplicates after removal:", df.duplicated().sum())

#7. check unique values in each column
print("Unique Values:", df.nunique())

#8. Fill numerical missing values with median
df["Item_price"] = df["Item_price"].fillna(df["Item_price"].median())

#9. Fill categorical missing values with mode
df["Customer_City"] = df["Customer_City"].fillna(df["Customer_City"].mode()[0])

#10. Display the first five rows
df.head()

````
